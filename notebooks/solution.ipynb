# Quantum Stone-Paper-Scissors: Experiments
Checks that the quantum choice is fair (1/3 each) and compares it with a classical random baseline.

```python
"""CIRCUIT REPRESENTATION"""

import sys; sys.path.append('../src')
from collections import Counter
import random
import matplotlib.pyplot as plt
from quantum_core import build_circuit, quantum_choice, get_winner

build_circuit().draw(output='mpl')
```

```python

"""QUANTITATIVE COMPARISON"""


N = 1500
quantum = Counter(quantum_choice() for _ in range(N))
classical = Counter(random.choice(['stone','paper','scissors']) for _ in range(N))
print('Quantum  :', dict(quantum))
print('Classical:', dict(classical))
```

```python

"""DIAGRAMATIC REPRESENTATION"""


moves = ['stone','paper','scissors']
plt.bar([i-0.2 for i in range(3)], [quantum[m]/N for m in moves], 0.4, label='Quantum')
plt.bar([i+0.2 for i in range(3)], [classical[m]/N for m in moves], 0.4, label='Classical')
plt.xticks(range(3), moves); plt.axhline(1/3, color='k', ls='--'); plt.legend(); plt.show()
```

```python

"""CHECKING FOR BIAS"""

# Win rate of a fixed-strategy player (always 'stone') vs quantum computer
res = Counter(get_winner('stone', quantum_choice()) for _ in range(N))
print({k: v/N for k, v in res.items()})
```
